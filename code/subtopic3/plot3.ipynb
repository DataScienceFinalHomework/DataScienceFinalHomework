## 子主题3绘图

```python
from plotnine import *
import pandas as pd
```

```python
data_path = "../../Datasets/subtopic3/subtopic3.csv"
data = pd.read_csv(data_path).convert_dtypes()
data = data.dropna()
data["dream_type"] = data["positive_or_not"].map({
    1: "positive_dream",
    0: "negative_dream",
})
data.head(3)
```

```python
def MyBoxPlot(y_axis):
    p = (
        ggplot(data, aes(x="dream_type", y=y_axis))
        + geom_boxplot()
        + labs(x="dream type", y=y_axis, title=y_axis+" boxplot")
    ) 
    return p
```

```python
# plot 箱线图

p_gini = MyBoxPlot("gini_coeff")
p_gini.save("gini_coeff_box_plot.jpg")

p_cluster = MyBoxPlot("cluster_coeff")
p_cluster.save("cluster_coeff_box_plot.jpg")

# avg_min_path: 限制y轴范围 (0, 10)，99%分位数在9.77
p_avg = (
    ggplot(data, aes(x="dream_type", y="avg_min_path"))
    + geom_boxplot()
    + scale_y_continuous(limits=(0, 10))
    + labs(x="dream type", y="avg_min_path", title="avg_min_path boxplot")
)
p_avg.save("avg_min_path_box_plot.jpg")

# feedback_loops: 限制y轴范围 (0, 50)，99%分位数在47
p_feed = (
    ggplot(data, aes(x="dream_type", y="feedback_loops"))
    + geom_boxplot()
    + scale_y_continuous(limits=(0, 50))
    + labs(x="dream type", y="feedback_loops", title="feedback_loops boxplot")
)
p_feed.save("feedback_loops_box_plot.jpg")
```

```python
# 数值指标对比表
metrics = ['cluster_coeff', 'feedback_loops', 'gini_coeff']
summary = data.groupby('dream_type')[metrics].agg(['mean', 'median', 'std']).round(4)
summary.to_csv("compare_table.csv", index=False, encoding="utf-8")
summary
```
